```python
subject1_marks = int(input("Enter the marks for subject 1: "))
subject2_marks = int(input("Enter the marks for subject 2: "))
subject3_marks = int(input("Enter the marks for subject 3: "))
subject4_marks = int(input("Enter the marks for subject 4: "))
subject5_marks = int(input("Enter the marks for subject 5: "))

average = (subject1_marks + subject2_marks + subject3_marks + subject4_marks + subject5_marks) / 5
if average >= 80:
    print("A")
elif average >= 60:
    print("B")
elif average >= 40:
    print("C")
else:
        print("F")
```

```python


def add(for_nunber1, for_nunber2):
    sum_of_numbers = for_nunber1 + for_nunber2
    return sum_of_numbers

def subtract(for_nunber1, for_nunber2):
    difference_of_numbers = for_nunber1 - for_nunber2
    return difference_of_numbers

def multiply(for_nunber1, for_nunber2):
    product_of_numbers = for_nunber1 * for_nunber2
    return product_of_numbers

def divide(for_nunber1, for_nunber2):
    quotient_of_numbers = for_nunber1 / for_nunber2
    return quotient_of_numbers

def calculator():
    user_cal = int(input("Enter number between 1 and 4 to start the calculator: "))
    for_nunber1 = int(input("Enter your first number: "))
    for_nunber2 = int(input("Enter your second number: "))
    if user_cal == 1:
        print(f"The sum is {add(for_nunber1, for_nunber2)}")
    elif user_cal == 2:
        print(f"The difference is {subtract(for_nunber1, for_nunber2)}")
    elif user_cal == 3:
        print(f"The product is {multiply(for_nunber1, for_nunber2)}")
    elif user_cal == 4:
        print(f"The quotient is {divide(for_nunber1, for_nunber2)}")
calculator()


```

```python
your_cart = {}

while True:
    item = input("Enter an item to add to your cart (or type 'check out' if finished): ")
    if item == 'check out':
        break
    price_tag = float(input(f"Enter the price of {item}: "))
    your_cart[item] = (price_tag)

print(your_cart)

total_price = 0
for price in your_cart.values():
    total_price = total_price + price
print(f"Total price: {total_price:.2f}")

```

```python
import random

while True:
    guessing_number = int(input("Guess a number between 1 and 20: "))
    random_number = random.randint(1, 20)
    if guessing_number == random_number:
        print("You are such a genuis, you guessed the right number")
        break
    elif guessing_number < random_number:
        print("You guessed too low, try again you can do it!")
    elif guessing_number > random_number:
        print("You guessed too high, try something lower!")
        
```

```python
student_name = []

while True:
    name = input("Enter student's name (or type 'done' if name is complete): ")
    if name == 'done':
        break
    student_name.append(name)

with open("student_name.txt", "w") as file:
    for name in student_name:
        file.write(str(name) + "\n")

with open("student_name.txt", "r") as file:
    names = [name.strip() for name in file.readlines()]
    print(names)
```

```python
averages = [68, 78, 90, 2, 10, 80, 91]

def the_largest_number(averages):
    largest = averages[0]

    for number in averages:
        if number > largest:
            largest = number
    return largest
print(the_largest_number(averages))
```
